## 3. Add/Modify a glycan manually

Use this helper when adding a glycan that is not yet in `config/glycans.yaml`. Only parameters that differ from the global defaults need to be supplied. The function validates the definition before writing it to the YAML file.

**Tip:** keep the residue/linkage order exactly as used by the GlyCONFORMER labelling scheme.


```python
 add_glycan_to_yaml(
     CONFIG_FILE,
     name='PolySia_DP3',
     path='data/Polymer/PolySia_DP3/',
     label=[[1, 2], [2, 3], [2, 3]],
     ltype=[8, 8, 8],
     carbon_t=[2, 2, 2],
     carbon_p=[2, 2, 2],
     modify_forcefield_for_neu5ac='true',
     peptide='none',
     monosaccharide_types='ANE5AC'
 )
```

## Execute

```python
selected_glycans = setup.available_glycans(config)
print(selected_glycans)
```

```python
for glycan in selected_glycans:
    setup.preview_setup(glycan, config, repo_root=REPO_ROOT)
    setup.glygen(
        glycan,
        config,
        repo_root=REPO_ROOT,
        template_dir=TEMPLATE_DIR,
        use_module_load=True,
        dry_run=False
    )
```
